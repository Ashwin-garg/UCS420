## Q1

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Temperature readings:", temperature)
```

### a)

```python
corrected_temperature = temperature + 2
print("Corrected temperature:", corrected_temperature)
```

### b)

```python
fahrenheit = (corrected_temperature * 9 / 5) + 32
print("Temperature in Fahrenheit:", fahrenheit)
```

### c)

```python
greater_than_32 = temperature[temperature > 32]
print("Readings greater than 32:", greater_than_32)
```

### d)

```python
count = np.sum(temperature > 32)
print("Number of readings greater than 32:", count)
```

### e)

Vectorization applies an operation on the whole array at once, so we do not need to write a loop for every value. Boolean indexing helps filter only the required values directly from the array. This makes the code shorter, cleaner, and faster for large data.

## Q2

```python
steps = np.array([[5000, 6200, 7100],
                  [8000, 7500, 9000],
                  [4500, 5100, 4800],
                  [9000, 8500, 9500]])
print(steps)
```

### a)

```python
total_steps = np.sum(steps)
print("Total steps:", total_steps)
```

### b)

```python
mean_steps = np.mean(steps)
print("Mean steps:", mean_steps)
```

### c)

```python
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))
```

### d)

```python
day_total = np.sum(steps, axis=0)
print("Total steps for each day:", day_total)
```

### e)

```python
user_total = np.sum(steps, axis=1)
print("Total steps for each user:", user_total)
```

### f)

```python
max_position = np.argmax(steps)
row, column = np.unravel_index(max_position, steps.shape)

print("Position of maximum value:", row, column)
print("Maximum value:", steps[row, column])
```

## Q4

```python
X = np.array([[6, 70, 3],
              [5, 50, 6],
              [8, 80, 2],
              [4, 30, 8]])

y = np.array([40, 65, 30, 85])

print("X:\n", X)
print("y:", y)
```

### a)

```python
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
```

### b)

```python
x_transpose = X.T
print("Transpose of X:\n", x_transpose)
```

The transpose changes rows into columns and columns into rows. Here, the features become rows.

### c)

```python
xtx = X.T @ X
print("X.T @ X:\n", xtx)
```

### d)

```python
xtx_inverse = np.linalg.inv(xtx)
print("Inverse of X.T @ X:\n", xtx_inverse)
```

### e)

```python
beta = xtx_inverse @ X.T @ y
print("Beta values:", beta)
```

### f)

The beta values are the coefficients for sleep hours, activity level, and stress level. They show how each feature contributes to the predicted assistance score.

### g)

```python
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
