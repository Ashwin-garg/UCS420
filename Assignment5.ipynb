### b) i)

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

count = np.bincount(x)
most_frequent = np.where(count == np.max(count))[0]

print("Most frequent value:", most_frequent)
for i in most_frequent:
    print("Indices of", i, ":", np.where(x == i)[0])
```

### b) ii)

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

count = np.bincount(y)
most_frequent = np.where(count == np.max(count))[0]

print("Most frequent value:", most_frequent)
for i in most_frequent:
    print("Indices of", i, ":", np.where(y == i)[0])
```

## Q4

```python
ashwin_garg = np.linspace(10, 100, 25)
print("Array:", ashwin_garg)
```

### a)

```python
print("Dimensions:", ashwin_garg.ndim)
print("Shape:", ashwin_garg.shape)
print("Total elements:", ashwin_garg.size)
print("Data type:", ashwin_garg.dtype)
print("Total bytes consumed:", ashwin_garg.nbytes)
```

### b)

```python
transpose_using_reshape = ashwin_garg.reshape(25, 1)
print("Transpose using reshape():\n", transpose_using_reshape)
print("Shape after reshape():", transpose_using_reshape.shape)
```

### c)

No. For a one-dimensional array, `T` does not change its shape. The array must first be reshaped to 2-D.

```python
transpose_with_T = ashwin_garg.T
print("Shape using T directly:", transpose_with_T.shape)

column_using_T = ashwin_garg.reshape(1, 25).T
print("Shape after reshaping to 2-D and using T:", column_using_T.shape)
```

## Q5

```python
nparr = np.array([[10, 20, 30, 40],
                              [50, 60, 70, 80],
                              [90, 15, 20, 35]])
print("2-D array:\n", nparr)
```

### a)

```python
print("Mean:", np.mean(nparr))
print("Median:", np.median(nparr))
print("Maximum:", np.max(nparr))
print("Minimum:", np.min(nparr))
print("Unique elements:", np.unique(nparr))
```

### b)

```python
reshaped_nparr = nparr.reshape(4, 3)
print("Reshaped array (4 x 3):\n", reshaped_nparr)
```

### c)

```python
resized_nparr = np.resize(nparr, (2, 3))
print("Resized array (2 x 3):\n", resized_nparr)
```
