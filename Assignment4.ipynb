## Q1

```python
roll_number = "1024170191"
digits = [int(i) for i in roll_number]
last_two_digits = digits[-2:]

print("Roll number:", roll_number)
print("Digits:", digits)
print("Last two digits:", last_two_digits)
```

```python
fixed_entries = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.", "keywords": "fee cost price charge", "category": "billing"},
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.", "keywords": "password reset login", "category": "account"},
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.", "keywords": "hours timing open time", "category": "general"},
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.", "keywords": "pay payment upi fee", "category": "billing"}
]

categories = ["billing", "account", "general"]
```

```python
personal_entries = [
    {
        "question": "how do i check my fee receipt",
        "answer": "You can check your fee receipt from the student payment portal.",
        "keywords": "receipt fee portal",
        "category": categories[last_two_digits[0] % 3]
    },
    {
        "question": "how do i update my registered mobile number",
        "answer": "You can update your mobile number from account settings.",
        "keywords": "mobile update account",
        "category": categories[last_two_digits[1] % 3]
    }
]

faq_data = fixed_entries + personal_entries
df = pd.DataFrame(faq_data)
print(df)
```

## Q2

```python
def score_query(query, df):
    query_words = query.lower().split()
    results = []
    
    for index, row in df.iterrows():
        text = row["question"] + " " + row["keywords"] + " " + row["category"]
        text = text.lower()
        score = 0
        
        for word in query_words:
            if word in text:
                score = score + 1
        
        if score > 0:
            results.append([index, row["question"], row["answer"], row["category"], score])
    
    result_df = pd.DataFrame(results, columns=["index", "question", "answer", "category", "score"])
    result_df = result_df.sort_values(by="score", ascending=False)
    return result_df

matched_entries = score_query("fee payment", df)
print(matched_entries)
```

## Q3

```python
def same_category(category_name, df):
    return df[df["category"] == category_name]["question"]

personal_category = personal_entries[0]["category"]
questions = same_category(personal_category, df)

print("Category:", personal_category)
print(questions)
```

## Q6

```python
def score_query_with_tie(query, df):
    query_words = query.lower().split()
    scores = []
    
    for index, row in df.iterrows():
        text = row["question"] + " " + row["keywords"] + " " + row["category"]
        text = text.lower()
        score = 0
        
        for word in query_words:
            if word in text:
                score = score + 1
        
        scores.append(score)
    
    df_copy = df.copy()
    df_copy["score"] = scores
    max_score = df_copy["score"].max()
    
    if max_score == 0:
        print("No matching entry found")
    else:
        best_matches = df_copy[df_copy["score"] == max_score]
        print(best_matches)

print("Query with tie:")
score_query_with_tie("fee", df)

print("\nQuery without tie:")
score_query_with_tie("password reset", df)
```
